```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Losowanie 
dane = np.random.normal(loc=100, scale=30, size=1000)

# histogram
plt.hist(dane, bins=20, edgecolor='black')
plt.title("Histogram wylosowanych danych")
plt.xlabel("Wartość")
plt.ylabel("Liczność")
plt.show()
```
